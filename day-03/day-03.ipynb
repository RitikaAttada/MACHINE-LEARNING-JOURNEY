# Lists And Tuples

### Lists:

```python
primes=[2,3,5,7]
Names=["Ritika", "Khyathi", "Sanjana", "Lalitha"]
print(primes)
print(Names)
```

### List of Lists:

```python
details = [
    ["Ritika", "Khyathi", "sanjana"],
    [10, 57, 58],
    ["Cse", "Csm", "Csd"]
]

print(details)
```

### Indexing:

```python
print(Names[1])
print(details[2][1])
print(Names[-1])
print(details[-1][-1])
```

### Slicing:

```python
# elements from index to count => slicing => Name[2:5]
print(Names[1: 3]) #the last is excluded
print(Names[:3])
print(Names[1:])
print(Names[:-1])
```

### Changing Lists:
#### Lists are mutable. To modify a list assign to the index or slice expression

```python
Names[3]="Sri Latha"
print(Names)
Names[0:2] = ["Sreeja", "krithi"]
print(Names)
```

```python
Names.append("Sivani") #adds at the back
print(Names)
```

### List functions:

```python
print(len(Names))
print(sorted(Names))
print(sum(primes))
print(max(primes))
```

# Loops and List Comprehensions

### Loops:

```python
for name in Names:
    print(name, end=' ')
```

```python
prod=1
for num in primes:
    prod = prod*num
print(prod)
```

```python
sentence = "Hello this is a random sentence to demonstrate the working of loops"

#print the letters followed by ' '
prev='0'
for ch in sentence:
    if prev==' ':
        print(ch, end=' ')
    prev=ch
    
```

### range():
#### this returns a sequence of numbers

```python
for i in range(5):
    print(i, end=" ")
```

### while loops:
#### iterates until the condition is met

```python
i=0
while i<10:
    print(i, end=" ")
    i+= 1
```

### List Comprehensions:

```python
squares = [n**2 for n in range(20)]
print(squares)
```

```python
planets = ["mercury", "venus", "earth", "mars", "jupiter", "saturn", "uranus", "neptune"]
short_planets = [p for p in planets if len(p)<6]
print(short_planets)
```

```python
nums=[5, -1, -2, 0, 3]
# count negatives
print(len([num for num in nums if num<0]))
```
