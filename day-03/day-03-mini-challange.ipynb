# 🧩 Day 03 — Mini Challenge: Planet Analyzer

## Given Data
```python
planets = [
    'Mercury', 'Venus', 'Earth', 'Mars',
    'Jupiter', 'Saturn', 'Uranus', 'Neptune'
]
```

## Objectives
Build a small program that:
1. Creates a list containing the length of each planet's name.
2. Finds the shortest planet name.
3. Finds the longest planet name.
4. Counts how many planet names have 5 or fewer characters.
5. Creates a new list containing the names of those short planets.

### 🔥 Bonus
* Do #1, #4 using list comprehensions.

---

## 🖥️ Output

```text
🪐 PLANET ANALYZER RESULTS 🪐
-----------------------------------
1. Name Lengths:       [7, 5, 5, 4, 7, 6, 6, 7]
2. Shortest Name:      Mars (4 chars)
3. Longest Name:       Mercury (7 chars)
4. Count (<= 5 chars): 3
```


````python
planets = [
    'Mercury', 'Venus', 'Earth', 'Mars',
    'Jupiter', 'Saturn', 'Uranus', 'Neptune'
]
````

````python
lengths = [len(planet) for planet in planets]
print(lengths)
````

````python
shortest = planets[0]
longest = planets[0]
for planet in planets:
    if len(planet)<len(shortest):
        shortest=planet
    elif len(planet)>len(longest):
        longest=planet

print(shortest)
print(longest)
````

````python
print(len([planet for planet in planets if len(planet)<6]))
````
