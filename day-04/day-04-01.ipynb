## Strings and Dictionaries

### Strings:
#### they can be defined using either single or double quotations. They are functionally equivalent.
#### Double quotes are convenient if your string contains a single quote character 
#### it's easy to create a string that contains double-quotes if you wrap it in single quotes

```python
print("Pluto's a planet")
print('My dog is named "pluto"')
```

```python
print('My dog's name is pluto')
```

##### we can fix the above problem by "escaping" the single quote using a backslah.

```python
print('My dog\'s name is pluto')
print("My dog's name is \"pluto\"")
```

##### \n - newline and \t - tabspace

```python
print("1\n2\t3")
```

##### triple quotes lets us include new lines by hitting "Enter" on our key board

```python
hello = "hello\nworld"
hello2 = """hello
world"""
print(hello)
print(hello2)
print(hello==hello2)
```

### strings are sequences

```python
planet="nepute"
print(planet[3])
print(planet[:-3])
print (len(planet))
```

```python
print([char+'*' for char in planet])
```

#### strings are immutable

```python
planet[0]='e'
```

### string methods
#### .upper() and .lower()

```python
sentence = "Hello people happy learning!"
print(sentence.upper())
print(sentence.lower())
print(sentence)
```

#### .index():
##### search for the first index of the given substring

```python
print(sentence.index("peop"))
```

```python
print(sentence.startswith("Hel"))
print(sentence.endswith("ing.")) # '.' at last isn't present in the original sentence
```

#### .split()

```python
words = sentence.split() #converts into a list of words breaking whitespace by default
print(words)
```

```python
date='2005-27-12'
year, month, day = date.split('-')
print(year, month, day)
```

#### .join()

```python
newDate = '/'.join([day, month, year])
print(newDate)
```

### Dictionaries:
##### These are python's built in data structure for mapping keys to values.
##### values are accesses via square bracket syntax similar to indexing

```python
numbers = {'one':1, 'two':2, 'three':3}
print(numbers['one'])
```

```python
numbers['four']=4
print(numbers)
numbers['five']=6
print(numbers)
#change 6 to 5 for numbers['five']
numbers['five']=5
print(numbers)
```

#### Dictionary Comprehensions:

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
map_planets = {planet: planet[0] for planet in planets}
print(map_planets)
```

##### a loop over a dictionary loops over its keys

```python
for k in numbers:
    print("{}={}".format(k, numbers[k]))
```
