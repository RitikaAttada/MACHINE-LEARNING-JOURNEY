### Arithematic operations

```python
# / - floating point division
# // -integer division

a=10
b=7
print(a/b, a//b)
```

```python
price=799
quantity=3
bill_before_discount = price*quantity
discount=15/100
final_bill = bill_before_discount*(1-discount)
print(final_bill)
```
