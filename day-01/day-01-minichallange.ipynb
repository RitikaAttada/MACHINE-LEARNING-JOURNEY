# BMI Calculator

Create a program that takes:

- Weight in kg
- Height in meters

and calculates:

$$
BMI = \frac{weight}{height^2}
$$

For example:

```python
weight = 60
height = 1.65

````python
weight_in_kg=58
height_in_meters=1.64

calculate_bmi = weight_in_kg/(height_in_meters**2)
print(calculate_bmi)
if (calculate_bmi<18.5):
    print("underweight")
elif (calculate_bmi<24.9):
    print("normal")
elif (calculate_bmi<30):
    print("overweight")
else:
    print("obese")
````
