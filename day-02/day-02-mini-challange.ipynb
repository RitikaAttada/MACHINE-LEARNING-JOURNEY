# Day-02 Mini Challenge: Simple Calculator

Create separate functions for:
* `add()`
* `subtract()`
* `multiply()`
* `divide()`

Then create a master function:
* `calculate(a, b, operation)`

The `operation` parameter determines which calculation to perform.

### Examples
* `calculate(10, 5, "add")`
* `calculate(10, 5, "multiply")`
* `calculate(10, 5, "divide")`

### Requirements
* Your calculator must handle **division by zero** properly without crashing.
* This project combines **functions**, **parameters**, **return values**


```python
def add(a, b):
    return a+b

def subtract(a, b):
    return a-b

def multiply(a, b):
    return a*b
    
def divide(a, b):
    if (b==0):
        return "zero division error: cannot divide a number by zero"
    return a/b
```

```python
def calculate(arg1, arg2, func):
    return func(arg1, arg2)
```

```python
print(calculate(10, 5, add))
print(calculate(10, 5, subtract))
print(calculate(10, 0, divide))
print(calculate(10,5, divide))
print(calculate(10, 5,  multiply))
```
