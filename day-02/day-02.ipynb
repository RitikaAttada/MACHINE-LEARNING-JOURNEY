### Defining functions:

```python
def square(num):
    return num*num

def cube(num):
    return square(num)*num

def calculate_area(length, width):
    return length*width

def average(a, b, c):
    return (a+b+c)/3

def largest(a, b, c):
    if (a>b and a>c):
        return a
    elif (b>a and b>c):
        return b
    return c
```

```python
num=3
print(square(num))
print(cube(num))
print(calculate_area(22/7, 5))
print(average(1,2,3))
print(largest(21, 16.8, 23.97))
```

### Docstring:
#### It is a triple-quoted string that comes after the header of a function. Ehen we call help() function, it shows the docstring

```python
def largest(a, b, c):
    """ Return the largest element
    """
    if (a>b and a>c):
        return a
    elif (b>a and b>c):
        return b
    return c
```

```python
help(largest)
```

### Void functions:
#### these functions dont return a value they just perform their intended task

```python
def largest(a, b, c):
    """ print the largest element
    """
    if (a>b and a>c):
        print(a)
    elif (b>a and b>c):
        print(b)
    print(c)
```

```python
largest( 1,2,3)
print(largest(1,2,3))
```

### Functions Applied to Functions:

```python
def twice(num):
    return (num*2)

def apply_function(func, arg):
    return func(arg)

def apply_function_twice(fun, arg):
    return fun(fun(arg))
    #if (fun, arg) is 6 this returns 12
```

```python
print(twice(3))
print(apply_function(twice, 6))
print(apply_function_twice(twice, 3))
```

### Challenge: Temperature Converter

Create two functions:
* `celsius_to_fahrenheit()`
* `fahrenheit_to_celsius()`

### Formulas
* **Celsius to Fahrenheit:** $F = (C \times \frac{9}{5}) + 32$
* **Fahrenheit to Celsius:** $C = (F - 32) \times \frac{5}{9}$

### Requirements
Both functions should:
* Accept a temperature as an argument.
* Return the converted temperature.
* Include a descriptive docstring.
* Be tested with multiple values.


```python
def celcius_to_fahrenheit(c):
    """
        Converts celcius temperature to Fahrenheit
    """
    f = (c*(9/5))+32
    return f


def fahrenheit_to_celcius(f):
    """
        Converts Fahrenheit temperature to celcuis
    """
    c = (f-32)*(5/9)
    return c
```

```python
help(celcius_to_fahrenheit)
help(fahrenheit_to_celcius)
```

```python
celcius_to_fahrenheit(78)
```

```python
fahrenheit_to_celcius(98)
```
